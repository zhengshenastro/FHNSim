# FHN Initial Condition & Parameter Exploration


This notebook summarizes recommended **initial conditions (ICs)** and **parameter sets** for:

- Regular FitzHugh–Nagumo (reaction–diffusion)
- Mass-conserved FitzHugh–Nagumo

Each section includes:
- Parameters
- Expected dynamics
- Physical interpretation


## 1. Pulse Initial Condition


### Regular FHN (R1 – Traveling Wave)

```python
a = 0.7
b = 0.8
epsilon = 0.08
Du = 1.0
Dv = 0.5
```

**Expected:**
- Circular outward propagating wave
- Clear wavefront in space-time plots

**Physics:**
- Classic excitable medium behavior

---

### Regular FHN (R2 – Failed Propagation)

```python
epsilon = 0.3
```

**Expected:**
- Pulse decays
- No sustained propagation

---

### Mass-Conserved FHN (MC1)

```python
Du = 1000
Dv = 0.5
```

**Expected:**
- No wave propagation

**Physics:**
- Laplacian suppresses excitability
